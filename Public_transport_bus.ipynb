<h1 style="text-align:left;">📖 Introduction</h1>

Public transportation plays an important role in providing affordable and convenient travel facilities to people. 

Buses are one of the most commonly used modes of public transportation and generate large amounts of data related to routes, passengers, trips, timings, distance, and other operational factors.

This project focuses on analyzing public transport bus data to understand passenger demand, route performance, travel patterns, and operational trends.

Data analytics techniques are used to clean, manipulate, analyze, and visualize the available bus transportation data.

<h1 style="text-align:left;">Problem Statement</h1>

Public transport systems generate large amounts of data every day.

Understanding this data manually can be difficult and time-consuming.

Without proper analysis, it can be challenging to identify busy routes, passenger demand, peak travel periods, and overall transportation patterns.

Therefore, this project aims to analyze public transport bus data to identify important patterns and generate useful insights that can help in understanding bus operations and passenger usage.

```python
delay = df.groupby("Traffic_Level")["Delay_Minutes"].mean()

plt.figure(figsize=(8,5), facecolor="lightyellow")

plt.bar(delay.index, delay.values,
        color=["green", "orange", "red"])

plt.title("Average Delay by Traffic Level")
plt.xlabel("Traffic Level")
plt.ylabel("Average Delay (Minutes)")

plt.grid(axis="y")

plt.show()
```

Q.Top 10 Routes uses by passengers?

```python
routes = df.groupby("Route_Name")["Passengers"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5), facecolor="mistyrose")

plt.bar(routes.index, routes.values,
        color=["red", "orange", "yellow", "green", "blue",
               "purple", "pink", "brown", "cyan", "gray"])

plt.title("Top 10 Routes by Passengers")
plt.xlabel("Route")
plt.ylabel("Passengers")

plt.xticks(rotation=45)

plt.grid(axis="y")
plt.show()
```

```python
revenue = df.groupby("Bus_Type")["Revenue"].sum()

plt.figure(figsize=(8,5), facecolor="lavender")

plt.bar(revenue.index, revenue.values,
        color=["teal", "gold", "tomato", "blue"])

plt.title("Revenue by Bus Type")
plt.xlabel("Bus Type")
plt.ylabel("Revenue")

plt.grid(axis="y")
plt.show()
```

```python
bus_type = df["Bus_Type"].value_counts()

plt.figure(figsize=(7,7), facecolor="lavender")

plt.pie(bus_type.values,
        labels=bus_type.index,
        autopct="%1.1f%%",
        colors=["blue", "orange", "green", "purple"],
        startangle=90)

plt.title("Bus Type Distribution")
plt.show()
```

```python
fuel_trend = df.groupby("Date")["Fuel_Consumption_Liters"].sum()

plt.figure(figsize=(10,5), facecolor="beige")

plt.plot(fuel_trend.index,
         fuel_trend.values,
         color="brown",
         linewidth=2)

plt.title("Fuel Consumption Over Time")
plt.xlabel("Date")
plt.ylabel("Fuel Consumption (Liters)")

plt.grid(True)
plt.show()
```

```python
status = df["Trip_Status"].value_counts()

plt.figure(figsize=(7,7), facecolor="lightyellow")

plt.pie(status.values,
        labels=status.index,
        autopct="%1.1f%%",
        colors=["green", "orange", "red"],
        startangle=90)

plt.title("Trip Status Distribution")
plt.show()
```

```python
plt.figure(figsize=(9,5), facecolor="mistyrose")

plt.hist(df["Delay_Minutes"],
         bins=15,
         color="purple",
         edgecolor="black")

plt.title("Distribution of Bus Delays")
plt.xlabel("Delay in Minutes")
plt.ylabel("Number of Trips")

plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(9,5), facecolor="lightcyan")

plt.scatter(df["Distance_km"],
            df["Passengers"],
            color="darkblue",
            alpha=0.6)

plt.title("Distance vs Passengers")
plt.xlabel("Distance (km)")
plt.ylabel("Passengers")

plt.grid(True,)
plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"])

passenger_trend = df.groupby("Date")["Passengers"].sum()

plt.figure(figsize=(10,5), facecolor="lightyellow")

plt.plot(passenger_trend.index,
         passenger_trend.values,
         color="green",
         marker="o",
         linewidth=2)

plt.title("Passenger Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Total Passengers")

plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
rating = df.groupby("Bus_Type")["Customer_Rating"].mean()

plt.figure(figsize=(8,5), facecolor="yellow")

plt.bar(rating.index, rating.values,
        color=["teal", "orange", "purple", "green"])

plt.title("Average Customer Rating by Bus Type")
plt.xlabel("Bus Type")
plt.ylabel("Average Rating")

plt.grid(axis="y")
plt.show()
```

```python
plt.figure(figsize=(9,5), facecolor="beige")

plt.scatter(df["Distance_km"],
            df["Revenue"],
            color="crimson",
            alpha=0.6)

plt.title("Distance vs Revenue")
plt.xlabel("Distance (km)")
plt.ylabel("Revenue")

plt.grid(True, alpha=0.5)

plt.show()
```

```python
weather = df["Weather"].value_counts()

plt.figure(figsize=(7,7), facecolor="lightblue")

plt.pie(weather.values,
        labels=weather.index,
        autopct="%1.1f%%",
        colors=["gold", "skyblue", "gray", "lightgreen"],
        startangle=90)

plt.title("Weather Condition Distribution")

plt.show()
```
